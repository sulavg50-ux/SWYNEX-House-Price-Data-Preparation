### Data Cleaning Assumptions

1. Missing numerical values are replaced using the median.
2. Missing categorical values are replaced using the most frequent value (mode).
3. Duplicate records are removed to avoid repeated data.
4. Original column names are preserved.
5. Numerical columns retain their numerical data types.
6. Categorical columns are converted into string format.
7. No unusual values or outliers are automatically removed.
8. The original dataset is preserved, and the cleaned dataset is saved separately.
9. Missing target values should be removed before supervised machine learning if a target column is used.

```python
# Identify numerical and categorical columns
numerical_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(include=['object']).columns

for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values using mode
for col in categorical_columns:
    mode_value = df[col].mode()
    if not mode_value.empty:
        df[col] = df[col].fillna(mode_value[0])
    else:
        df[col] = df[col].fillna("Unknown")
print("Missing values handled successfully!")
```

```python
for col in categorical_columns:
    df[col]=df[col].astype('string').str.strip()

display(df.dtypes)
```
